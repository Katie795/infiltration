## Running an infiltration simulation

The default geometrical configurations for the rain garden simulation are shown in the picture below

![rain_garden_domain](rain_garden_domain.png)

These lengths can be changed in the `config_rain_garden.toml` file in the `scripts/` folder.

The Gamma symbols correspond to different types of boundary:
- Gamma_n corresponds to the surface of the soil where there is a flux of water going through the boundary. This flux corresponds to the rainfall rate. If ponding conditions occur, The value on Gamma_n is set to be equal to the height of the ponded water
- Gamma_i corresponds to impervious boundary condtions (zero flux boundary conditions)
- Gamma_s corresponds to seepage boundary conditons. When the soil is saturated, there is zero flux out of the domain, but when the soil is saturated, there is flux.


The parameters in `infiltration_simulation.py` can be changed via `config_rain_garden.toml`. This file looks like

### Configuration file for Malago road simulation
```
[geometry]
length = 4         # length of the rain garden (m)
depth  = 1.2       # depth of the rain garden (m)
drain_l = 1.775    # x-coordinate of left hand side drain (m)
drain_r = 2.225    # y-coordinate of right hand side of drain (m)

[mesh]
nx = 255           # number of elements in the x-direction
ny = 63            # number of elements in the y-direction

[time]
dt = 0.5          # timestep (s)
T = 86400            # total length of time (s)
nt = 172800        # total number of time-steps

[IO]
rainfall_data =    "../data/Clifton_rain_modifed_for_rainfall_event.csv" # csv file containing rainfall data
save_fig = true    # whether to save the results to be post-processed
n_out = 100        # how often to save results (every n_out timesteps)

[simulation]
device = "cuda:0"     # which device to run the simulation on. Options: cpu, cuda:0
surface = "rainfall" # which type of boundary conditions to use on the surface of the soil. Options: "ponding" or "rainfall"
model = "L" # which model to use. Options: "L" or "PC"
```

These options can be changed to modify the solver.

Note that if you are using the PC method, then nx + 1 and ny + 1 should be divisible be divisible by 8 in order for the multigrid method to work.

There are many options for outputs. Running
```
python create_plots.py --potential --nfig=70
```
gives an animation of the evolution of the potential, and 

```
python create_plots.py --flux --nfig=70
```
gives an animation of the water flux.


## Other types of simulations

The `swe_simulation.py` runs the shallow water equations model.

The `horung_messing_test.py` and `horung_messing_FEM_test.py` runs the Horung-Messing, which is a problem where the analytical solution is known. The `horung_messing_FEM_test.py` is run using [Firedrake](https://www.firedrakeproject.org/) and requires a different virtual environment



The code solves the Richards equation

$$ \frac{\partial \theta(u)}{\partial t} - \nabla \cdot (\kappa(u)\nabla u) = f $$
