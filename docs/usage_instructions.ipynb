# Usage instructions

More details about the inner workings of the code can be found at the bottom of this document.



## Running an infiltration simulation

The default geometrical configurations for the rain garden simulation are shown in the picture below

![rain_garden_domain](rain_garden_domain.png)

These lengths can be changed in the `config_rain_garden.toml` file in the `scripts/` folder.

The Gamma symbols correspond to different types of boundary:
- Gamma_n corresponds to the surface of the soil where there is a flux of water going through the boundary. This flux corresponds to the rainfall rate. If ponding conditions occur, The value on Gamma_n is set to be equal to the height of the ponded water
- Gamma_i corresponds to impervious boundary condtions (zero flux boundary conditions)
- Gamma_s corresponds to seepage boundary conditons. When the soil is saturated, there is zero flux out of the domain, but when the soil is saturated, there is flux.


### Solver configurations

The parameters in `infiltration_simulation.py` can be changed via `config_rain_garden.toml`. This file looks like

```
[geometry]
length = 4         # length of the rain garden (m)
depth  = 1.2       # depth of the rain garden (m)
drain_l = 1.775    # x-coordinate of left hand side drain (m)
drain_r = 2.225    # y-coordinate of right hand side of drain (m)

[mesh]
nx = 255           # number of elements in the x-direction
ny = 63            # number of elements in the y-direction

[time]
dt = 0.5           # timestep (s)
T = 3600           # total length of time (s)
nt = 7200          # total number of time-steps

[IO]
rainfall_data =    "../data/Clifton_rain_modifed_for_rainfall_event.csv" # csv file containing rainfall data
save_fig = true    # whether to save the results to be post-processed
n_out = 100        # how often to save results (every n_out timesteps)

[simulation]
device = "cpu"     # which device to run the simulation on. Options: cpu, cuda:0
surface = "rainfall" # which type of boundary conditions to use on the surface of the soil. Options: "ponding" or "rainfall"
model = "L" # which model to use. Options: "L" or "PC"
```
Some things to note:
- The solver has two possible methods: the L-method and the PC method. The L-method is more stable but might take longer to run, while the PC method is faster (as it uses a multigrid method). 
- I you are using the PC method, then nx + 1 and ny + 1 should be divisible be divisible by 8 in order for the multigrid method to work.
- In the `[time]` parameters, you must have $nt \times dt = T$.
- In the `[IO]` (input-output) parameters, `rainfall_data` is the path to your csv file containing rainfall data. If this is not in 15 minute intervals, then the function `rainfall_flux()` in `infiltration_simulation.py` will need to be modified.
- To run the simulation on a GPU, use the `cuda:0` option in the `[simulation]` paurameters.
- If you are using the `ponding` surface option in the `[simulation]` parameters, and you want to run a simulation without the default ponding levels, you will need to modify `g()` in `infiltration_simulation.py`


### Post-processing

Running
```
python create_plots.py --potential --nfig=70
```
will give an animation of the evolution of the potential, and 

```
python create_plots.py --flux --nfig=70
```
gives an animation of the water flux. If you want to save the png files themselves and not produce an animation, you will need to modify the `create_plots.py` by commenting out the relevant lines of code.


## Other simulations
The `horung_messing_test.py` and `horung_messing_FEM_test.py` in the `numerical_experiments/` folder, runs the Horung-Messing, which is the problem where the analytical solution is known. The `horung_messing_FEM_test.py` is run using [Firedrake](https://www.firedrakeproject.org/) and requires a different virtual environment.

The `swe_simulation.py` runs the shallow water equations model. This is for future work, so this hasn't been thoroughly tested.

The `tests/` folder contains some pytest tests to check whether certain aspects of the code are working correctly. These can be run by command
```
pytest tests/
```
